```python
data =[
    (1 , "Parth", 22, 110000),
    (2 , "Vishwa", 20, 95000),
    (3 , "Kartik", 22, 250000)
]

columns = ["id", "name", "age","clg_fees"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
df.printSchema()
```

```python
df.select("name","age","clg_fees").show()
```

```python
df.filter(df.age < 21).show()
```

```python
from pyspark.sql import functions as F

df_with_fees = df.withColumn(
    "4years_fees", F.col("clg_fees") * 4)

df_with_fees.show()

```
